**Analysing the Dataset**

```python
import numpy as np
import matplotlib.pyplot as plt

# Load images and labels
images = np.load("data/images.npy")  
labels = np.load("data/labels.npy")

# Check shape
print(f"Images shape: {images.shape}")  
print(f"Labels shape: {labels.shape}")

# Display a random image
index = np.random.randint(0, len(images))  
plt.imshow(images[index], cmap="gray" if images.ndim == 3 else None)  
plt.title(f"Label: {labels[index]}")
plt.axis("off")
plt.show()
```

```python
# Load labels
labels = np.load("data/labels.npy")

# Count occurrences of each label
count_labels = np.bincount(labels)

# Display results
print(f"Total images with Label 0: {count_labels[0]}")
print(f"Total images with Label 1: {count_labels[1]}")
```

**Plotting Images With Their Corresponding Labels**

```python
images = np.load("data/images.npy")
labels = np.load("data/labels.npy")

unique_labels = np.unique(labels)

fig, axes = plt.subplots(len(unique_labels), 5, figsize=(10, 10))
for i, label in enumerate(unique_labels):
    indices = np.where(labels == label)[0][:5]
    for j, idx in enumerate(indices):
        axes[i, j].imshow(images[idx], cmap="gray")
        axes[i, j].set_title(f"Label: {label}")
        axes[i, j].axis("off")
plt.show()
```

**Pixel Intensity Distribution**

```python
plt.figure(figsize=(8, 5))
plt.hist(images.flatten(), bins=50, color="blue", alpha=0.7)
plt.title("Pixel Intensity Distribution")
plt.xlabel("Pixel Value (0-255 or 0-1)")
plt.ylabel("Count")
plt.show()
```
